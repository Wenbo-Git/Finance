```python
import pandas as pd
import numpy as np
from io import StringIO
```

```python
# --- prices.csv ---
prices_data = StringIO("""
ts,symbol,price
2026-02-20 09:30:00,AAPL,190.00
2026-02-20 09:31:00,AAPL,190.40
2026-02-20 09:32:00,AAPL,189.80
2026-02-20 09:30:00,TSLA,205.00
2026-02-20 09:31:00,TSLA,209.10
2026-02-20 09:32:00,TSLA,206.00
2026-02-20 09:30:00,SPY,500.00
2026-02-20 09:31:00,SPY,499.50
2026-02-20 09:32:00,SPY,499.75
""")

prices = pd.read_csv(prices_data, parse_dates=["ts"])

# --- positions.csv ---
positions_data = StringIO("""
account_id,symbol,qty,avg_cost
1001,AAPL,200,185.00
1001,SPY,-50,505.00
1002,TSLA,30,210.00
1003,AAPL,10,195.00
""")

positions = pd.read_csv(positions_data)

# --- margin.csv ---
margin_data = StringIO("""
symbol,margin_rate
AAPL,0.30
TSLA,0.60
SPY,0.15
""")

margin = pd.read_csv(margin_data)
```

```python
prices
```

```python
prices=prices.sort_values(["symbol","ts"])
```

```python
prices["log_ret"]=np.log(prices["price"]/prices.groupby("symbol")["price"].shift(1))
```

```python
prices
```

```python
df_vol=prices.groupby("symbol")["log_ret"].apply(lambda x:np.sqrt(np.sum(x**2))).reset_index(name="realized_vol")
```

```python
df_vol
```

```python
df_vol["flag"]=np.where(df_vol["realized_vol"]>0.015, "High Volatility", "Low Volatility")
```

```python
df_vol
```

```python
df_var=prices.groupby("symbol")["log_ret"].quantile(0.01).reset_index(name="var_99")
```

```python
df_var["var_99"]=df_var["var_99"]*(-1)
```
